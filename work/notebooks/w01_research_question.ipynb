## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the FlyRank starter dataset from our GitHub repository
url = "https://raw.githubusercontent.com/Shrinidhi-02/flyrank-machine-learning/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

# Basic dataset size
total_pages = len(df)

# Pages with enough age and search visibility to be considered for refresh
refresh_candidates = df[
    (df["content_age_days"] >= 90) &
    (df["impressions_90d"] > 0)
].copy()

# Count pages showing a downward trend
declining_pages = (refresh_candidates["trend_direction"] == "down").sum()

# Percentage of refresh candidates that are declining
declining_rate = (
    declining_pages / len(refresh_candidates) * 100
    if len(refresh_candidates) > 0 else 0
)

print(f"Total pages: {total_pages:,}")
print(f"Pages eligible for refresh analysis: {len(refresh_candidates):,}")
print(f"Declining pages among these candidates: {declining_pages:,}")
print(f"Declining rate: {declining_rate:.2f}%")
```

The starter dataset contains 30,000 pages. After applying the refresh-analysis filters of at least 90 days of content age and more than zero impressions in the last 90 days, all 30,000 pages remain eligible. Among these pages, 16,262 are marked as declining, which is 54.21% of the dataset. These numbers show that there is a substantial group of pages whose performance may deserve further investigation, making refresh opportunity prioritization a useful direction for the next stages of the project.
